```python
import pandas as pd
```

```python
fixed_entries = [
    {"question": "what is the annual fee",
     "answer": "The annual fee is Rs 500.",
     "keywords": "fee cost price charge",
     "category": "billing"},

    {"question": "how to reset password",
     "answer": "Go to Settings > Reset Password.",
     "keywords": "password reset login",
     "category": "account"},

    {"question": "what are your working hours",
     "answer": "We are open 9 AM to 5 PM.",
     "keywords": "hours timing open time",
     "category": "general"},

    {"question": "how can i pay the fee",
     "answer": "You can pay via UPI, card, or net banking.",
     "keywords": "pay payment upi fee",
     "category": "billing"}
]

personalized_entries = [
    {"question": "how can i check my bill",
     "answer": "You can check your bill in the billing section.",
     "keywords": "bill billing amount",
     "category": "billing"},

    {"question": "how do i get a payment receipt",
     "answer": "You can download the receipt from the billing section.",
     "keywords": "payment receipt download",
     "category": "billing"}
]

all_entries = fixed_entries + personalized_entries

df = pd.DataFrame(all_entries)

print(df)
```

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():
        text = row["question"] + " " + row["keywords"]
        text_words = set(text.lower().split())

        score = len(query_words & text_words)

        if score > 0:
            results.append((score, row["question"], row["answer"]))

    results.sort(reverse=True)

    for score, question, answer in results:
        print("Score:", score)
        print("Question:", question)
        print("Answer:", answer)
        print()
```

```python
score_query("how can i pay fee", df)
```

```python
def same_category(category_name, df):
    return df[df["category"] == category_name]["question"]

print(same_category("billing", df))
```

```python
def score_query(query, df):
    results = []

    for _, row in df.iterrows():
        text = row["question"] + " " + row["keywords"]
        score = sum(word in text.lower() for word in query.lower().split())

        if score > 0:
            results.append((score, row["question"], row["answer"]))

    max_score = max([x[0] for x in results])

    for score, question, answer in results:
        if score == max_score:
            print("Score:", score)
            print("Question:", question)
            print("Answer:", answer)
            print()


score_query("fee", df)
```
