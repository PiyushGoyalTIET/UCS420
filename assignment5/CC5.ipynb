```python
import numpy as np
```

```python
arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)

print("After adding 2:", arr + 2)

print("After multiplying by 3:", arr * 3)

print("After dividing by 2:", arr / 2)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])

print(arr[::-1])
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)

most_frequent = values[np.argmax(counts)]

print("Most frequent value:", most_frequent)
print("Indices:", np.where(x == most_frequent)[0])
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

most_frequent = values[np.argmax(counts)]

print("Most frequent value:", most_frequent)
print("Indices:", np.where(y == most_frequent)[0])
```

```python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print(arr[0, 1])
print(arr[2, 0])
```

```python
Piyush = np.linspace(10, 100, 25)

print("Array:", Piyush)
print("Dimensions:", Piyush.ndim)
print("Shape:", Piyush.shape)
print("Total elements:", Piyush.size)
print("Data type:", Piyush.dtype)
print("Total bytes:", Piyush.nbytes)

transpose_Piyush = Piyush.reshape(1, 25)
print("Transpose using reshape:")
print(transpose_Piyush)

print("Transpose using T:")
print(Piyush.T)
```

```python
ucs420_Piyush = np.array([[10, 20, 30, 40],
                          [50, 60, 70, 80],
                          [90, 15, 20, 35]])

print("Array:")
print(ucs420_Piyush)

print("Mean:", np.mean(ucs420_Piyush))
print("Median:", np.median(ucs420_Piyush))
print("Maximum:", np.max(ucs420_Piyush))
print("Minimum:", np.min(ucs420_Piyush))
print("Unique elements:", np.unique(ucs420_Piyush))

reshaped_ucs420_Piyush = ucs420_Piyush.reshape(4, 3)
print("Reshaped array:")
print(reshaped_ucs420_Piyush)

resized_ucs420_Piyush = np.resize(ucs420_Piyush, (2, 3))
print("Resized array:")
print(resized_ucs420_Piyush)
```
