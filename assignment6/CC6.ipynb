```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = np.linspace(-10, 10, 100)
M = int(input("Enter a value: "))
y1 = M*x**2
y2 = M*np.sin(x)

print(y1)
print(y2)
```

```python
plt.plot(x, y1, 'r--')
plt.plot(x, y2, 'g')
plt.grid()
plt.legend()
plt.title("Mathematical Functions")
plt.show()
```

```python
df = pd.DataFrame({
    'Subjects':['Maths', 'Science', 'English', 'History', 'Geography'],
    'Marks':[92, 98, 95, 100, 70]
})
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(data=df, x="Subjects", y="Marks", hue="Subjects", legend=False)

for i, value in enumerate(df["Marks"]):
    plt.text(i, value + 0.5, str(value), ha="center")

plt.title("Subject-wise Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

```python

np.random.seed(1024170406)
data = np.random.randn(50)

fig, ax = plt.subplots(2, 2, figsize=(10, 6))

ax[0, 0].plot(np.cumsum(data))
ax[0, 0].set_title("Cumulative Sum")
ax[0, 0].set_xlabel("Index")
ax[0, 0].set_ylabel("Sum")
ax[0, 0].grid()

ax[0, 1].scatter(range(50), data)
ax[0, 1].set_title("Random Noise")
ax[0, 1].set_xlabel("Index")
ax[0, 1].set_ylabel("Value")
ax[0, 1].grid()

ax[1, 0].plot(data)
ax[1, 0].set_title("Random Values")
ax[1, 0].grid()

ax[1, 1].hist(data)
ax[1, 1].set_title("Distribution")
ax[1, 1].grid()

plt.tight_layout()
plt.show()
```

```python
df = pd.read_csv("company_sales_data.csv")
```

```python
sns.lineplot(data=df, x="month_number", y="total_profit")

plt.title("Total Profit of All Months")
plt.xlabel("Month")
plt.ylabel("Total Profit")
plt.grid()
plt.show()
```

```python
products = ["facecream", "facewash", "toothpaste",
            "bathingsoap", "shampoo", "moisturizer"]

for product in products:
    sns.lineplot(data=df, x="month_number", y=product, label=product)

plt.title("Monthly Product Sales")
plt.xlabel("Month")
plt.ylabel("Units Sold")
plt.grid()
plt.show()
```

```python
sns.barplot(data=df.drop(columns="month_number"))

plt.title("Company Sales Data")
plt.xlabel("Attributes")
plt.ylabel("Values")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.show()
```
